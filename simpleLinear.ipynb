```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('/content/IceCreamData.csv')
print("dataset loaded")
```

```python
df.shape
```

```python
df.head()
```

```python
df.describe()
```

```python
df['Temperature']=df['Temperature'].apply(np.floor)
df
```

```python
df.isnull().sum()
```

```python
x=df.iloc[:,:-1]
x
```

```python
#dependent variable salary
y=df.iloc[:,-1]
y
```

```python
# Split the dataset into training and testing sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
#create the model
model=LinearRegression()
```

```python
#train the model
model.fit(x_train,y_train)
print("Model Trained")
```

```python
#generate predictionsfor trainint testingdataset using trained model
y_pred_train=model.predict(x_train)
y_pred_test=model.predict(x_test)
print("Predictions generated for both training and testing")
```

```python
#evaLUATION using diffrent metrics
print("Metrics for training dataset")
print("Mean Absolute Error:",mean_absolute_error(y_train,y_pred_train))
print("Mean Squared Error:",mean_squared_error(y_train,y_pred_train))
print("Root Mean Squared Error:",np.sqrt(mean_squared_error(y_train,y_pred_train)))
print("R2 Score:",r2_score(y_train,y_pred_train))
```

```python
plt.scatter(x_train,y_train,color='blue',label='Actual Training Datapoints')
plt.plot(x_train,y_pred_train,color='purple',linewidth=2,label='Fitted Line')
plt.xlabel('Temperature')
plt.ylabel('Revenue')
plt.title('Actual Training Datapoints vs Fitted Line')
plt.legend()
plt.show()
```
